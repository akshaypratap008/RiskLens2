```python
from dotenv import load_dotenv
import os
import requests
```

```python
load_dotenv()
```

```python
API_KEY = os.getenv("COMPANY_HOUSE_API")
COMPANY_NUMBER = "03639100"
URL = f"https://api.company-information.service.gov.uk/company/{COMPANY_NUMBER}"
```

```python
response = requests.get(
    url = URL,
    auth = (API_KEY, "")
)
```

```python
print(response.json())
```
