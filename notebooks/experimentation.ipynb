## Directors

```python
url = URL + "/officers"
url
```

```python
response = requests.get(
    url,
    auth= (API_KEY, "")
)
```

```python
officiers_data = response.json()
```

```python
officiers_data
```

```python
directors = []
secretary = []
others = []

for item in officiers_data['items']:
    if item['officer_role'] == "director":
        directors.append(
            {
                "name": item['name'],
                "appointment_date": datetime.strptime(item['appointed_on'], "%Y-%m-%d").date(),
                "nationality": item['nationality'],
                "tenure_in_years": round((date.today() - datetime.strptime(item['appointed_on'], "%Y-%m-%d").date()).days/365, 2)
            }
        )
    elif item['officer_role'] == "secretary":
        secretary.append(
            {
                'name': item['name'],
                "appointment_date": datetime.strptime(item['appointed_on'], "%Y-%m-%d").date(),
                "nationality": item['nationality'],
                "tenure_in_years": round((date.today() - datetime.strptime(item['appointed_on'], "%Y-%m-%d").date()).days/365, 2)
            }
        )
    else:
        others.append(
            {
                "name": item['name'],
                "appointment_date": datetime.strptime(item['appointed_on'], "%Y-%m-%d").date(),
                "nationality": item['nationality'],
                "role": item['officer_role']
            }
        )

directors
```

```python
officiers = {
    "directors": directors,
    "director_count": len(directors),
    "secretary": secretary,
    "avg_director_tenure": np.mean([director['tenure_in_years'] for director in directors])
}
```

```python
officiers
```
